# Cheatsheet: tablicni izvodi i integrali

Brzi notebook za ponavljanje najcescih izvoda i integrala. Fokus je na formulama koje se stalno koriste u probabilistici, RL/control zadacima, optimizaciji i generalnom racunu.

Konvencije: `C` je konstanta integracije, `a` je konstanta, i pretpostavi domenu gdje izrazi imaju smisla.

```python
import sympy as sp
x, a, b = sp.symbols('x a b', positive=True)
sp.init_printing()
```

## Osnovna pravila izvoda

| izraz | izvod |
|---|---|
| `c` | `0` |
| `x` | `1` |
| `x^n` | `n x^(n-1)` |
| `c f(x)` | `c f'(x)` |
| `f(x) + g(x)` | `f'(x) + g'(x)` |
| `f(x) g(x)` | `f'(x)g(x) + f(x)g'(x)` |
| `f(x)/g(x)` | `(f'(x)g(x) - f(x)g'(x))/g(x)^2` |
| `f(g(x))` | `f'(g(x)) g'(x)` |

Najcesca greska na ispitu: zaboraviti chain rule. Ako vidis nesto tipa `(3x+1)^5`, vanjski izvod puta unutrasnji izvod.

| `sin(x)` | `cos(x)` |
| `cos(x)` | `-sin(x)` |
| `tan(x)` | `sec^2(x) = 1/cos^2(x)` |
| `arcsin(x)` | `1/sqrt(1-x^2)` |

| `cosh(x)` | `sinh(x)` |

| `sigmoid(x)=1/(1+e^(-x))` | `sigmoid(x)(1-sigmoid(x))` |

## Osnovni integrali

| integral | rezultat |
|---|---|
| `int c dx` | `cx + C` |
| `int x^n dx`, `n != -1` | `x^(n+1)/(n+1) + C` |
| `int 1/x dx` | `ln|x| + C` |
| `int e^x dx` | `e^x + C` |
| `int e^(ax) dx` | `e^(ax)/a + C` |
| `int a^x dx` | `a^x/ln(a) + C` |
| `int sin(x) dx` | `-cos(x) + C` |
| `int cos(x) dx` | `sin(x) + C` |
| `int sec^2(x) dx` | `tan(x) + C` |
| `int 1/(1+x^2) dx` | `arctan(x) + C` |
| `int 1/sqrt(1-x^2) dx` | `arcsin(x) + C` |
| `int sinh(x) dx` | `cosh(x) + C` |
| `int cosh(x) dx` | `sinh(x) + C` |

## Cesti oblici sa linearnom unutrasnjom funkcijom

| integral | rezultat |
|---|---|
| `int (ax+b)^n dx`, `n != -1` | `(ax+b)^(n+1)/(a(n+1)) + C` |
| `int 1/(ax+b) dx` | `(1/a) ln|ax+b| + C` |
| `int e^(ax+b) dx` | `(1/a)e^(ax+b) + C` |
| `int sin(ax+b) dx` | `-(1/a)cos(ax+b) + C` |
| `int cos(ax+b) dx` | `(1/a)sin(ax+b) + C` |

Ovo je samo u-substitution: `u = ax+b`, `du = a dx`.

## Metode integracije koje treba prepoznati

Substitution:

`int f(g(x)) g'(x) dx = int f(u) du`, gdje je `u=g(x)`.

Integration by parts:

`int u dv = uv - int v du`.

Dobra heuristika za parts: LIATE redoslijed za `u`: Logarithmic, Inverse trig, Algebraic, Trig, Exponential.

## Gaussian i probability integrali

| integral | rezultat |
|---|---|
| `int_{-inf}^{inf} exp(-x^2) dx` | `sqrt(pi)` |
| `int_{-inf}^{inf} exp(-a x^2) dx`, `a>0` | `sqrt(pi/a)` |
| `int_{-inf}^{inf} exp(-(x-mu)^2/(2 sigma^2)) dx` | `sqrt(2 pi) sigma` |
| normal pdf integral | `int pdf(x) dx = 1` |
| standard normal second moment | `int x^2 phi(x) dx = 1` |

Za normalnu distribuciju `X ~ N(mu, sigma^2)`: `E[X]=mu`, `Var(X)=sigma^2`, `E[X^2]=sigma^2+mu^2`.

```python
# Brza provjera par formula u SymPy
examples = [
    sp.sin(3*x + 1),
    sp.exp(2*x),
    (2*x + 5)**4,
    sp.log(4*x + 2),
]

for f in examples:
    print("f(x)   =", f)
    print("d/dx   =", sp.diff(f, x))
    print("int dx =", sp.integrate(f, x))
    print()
```

## Mini drill

Probaj prvo sam, pa provjeri SymPyjem:

1. `d/dx (x^2 e^x)`
2. `d/dx ln(1+x^2)`
3. `int x e^(x^2) dx`
4. `int x cos(x) dx`
5. `int 1/(3x+2) dx`

```python
drill = [
    x**2 * sp.exp(x),
    sp.log(1+x**2),
    x * sp.exp(x**2),
    x * sp.cos(x),
    1/(3*x+2),
]

print("Derivatives:")
for f in drill[:2]:
    print(f, "->", sp.diff(f, x))

print("\nIntegrals:")
for f in drill[2:]:
    print(f, "->", sp.integrate(f, x))
```
